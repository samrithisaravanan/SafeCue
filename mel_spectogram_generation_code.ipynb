```python
import os
import librosa
import librosa.display
import matplotlib.pyplot as plt
import numpy as np


# ============================================================
# FOLDER LOCATIONS
# ============================================================

BASE_DIR = "/content/drive/Shareddrives/Safe Cue Project/SafeCue Dataset"

INPUT_DIR = os.path.join(
    BASE_DIR,
    "one_second_clips"
)

OUTPUT_DIR = os.path.join(
    BASE_DIR,
    "mel_spectrograms"
)


# ============================================================
# MEL SPECTROGRAM SETTINGS
# ============================================================

SAMPLE_RATE = 16000
N_MELS = 64


# ============================================================
# CHECK FOLDERS
# ============================================================

print("Input folder:", INPUT_DIR)
print("Input folder exists:", os.path.exists(INPUT_DIR))

if not os.path.exists(INPUT_DIR):
    raise FileNotFoundError(
        f"Could not find: {INPUT_DIR}"
    )

os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)


# ============================================================
# TRAIN / TEST / VALIDATION
# ============================================================

DATASETS = [
    "train",
    "test",
    "validation"
]

total_created = 0


for dataset in DATASETS:

    dataset_input = os.path.join(
        INPUT_DIR,
        dataset
    )

    # Check that dataset folder exists
    if not os.path.isdir(dataset_input):
        print(f"\nWARNING: {dataset_input} does not exist")
        continue

    # Create corresponding output folder
    dataset_output = os.path.join(
        OUTPUT_DIR,
        dataset
    )

    os.makedirs(
        dataset_output,
        exist_ok=True
    )

    print("\n======================================")
    print(f"Processing: {dataset}")
    print("======================================")


    # ========================================================
    # PROCESS EACH CLASS
    # ========================================================

    for class_name in os.listdir(dataset_input):

        class_input = os.path.join(
            dataset_input,
            class_name
        )

        # Only process class folders
        if not os.path.isdir(class_input):
            continue

        # Corresponding output class folder
        class_output = os.path.join(
            dataset_output,
            class_name
        )

        os.makedirs(
            class_output,
            exist_ok=True
        )

        print(f"\nClass: {class_name}")


        # ====================================================
        # FIND WAV FILES
        # ====================================================

        wav_files = [
            filename
            for filename in os.listdir(class_input)
            if filename.lower().endswith(".wav")
        ]

        print(f"Found {len(wav_files)} WAV files")


        # ====================================================
        # PROCESS WAV FILES
        # ====================================================

        for filename in wav_files:

            input_file = os.path.join(
                class_input,
                filename
            )

            try:

                # Load audio
                y, sr = librosa.load(
                    input_file,
                    sr=SAMPLE_RATE,
                    mono=True
                )


                # =================================================
                # GENERATE MEL SPECTROGRAM
                # =================================================

                mel_spectrogram = librosa.feature.melspectrogram(
                    y=y,
                    sr=sr,
                    n_mels=N_MELS
                )


                # Convert to decibels
                mel_spec_db = librosa.power_to_db(
                    mel_spectrogram,
                    ref=np.max
                )


                # =================================================
                # CREATE IMAGE
                # =================================================

                plt.figure(
                    figsize=(10, 4)
                )

                librosa.display.specshow(
                    mel_spec_db,
                    sr=sr,
                    x_axis="time",
                    y_axis="mel"
                )

                # Remove axes and labels
                plt.axis("off")

                # Remove whitespace
                plt.tight_layout(
                    pad=0
                )


                # =================================================
                # OUTPUT FILE
                # =================================================

                base_name = os.path.splitext(
                    filename
                )[0]

                output_file = os.path.join(
                    class_output,
                    base_name + ".png"
                )


                # Save image
                plt.savefig(
                    output_file,
                    bbox_inches="tight",
                    pad_inches=0,
                    dpi=300
                )

                plt.close()


                total_created += 1

                print(
                    f"  ✓ Created: {base_name}.png"
                )


            except Exception as e:

                print(
                    f"  ✗ ERROR processing {filename}: {e}"
                )


# ============================================================
# FINISHED
# ============================================================

print("\n======================================")
print("Finished creating Mel spectrograms!")
print(f"Total PNG files created: {total_created}")
print("======================================")
```
